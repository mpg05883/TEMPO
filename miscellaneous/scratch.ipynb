Define foo class

```python
class Foo:
    def __init__(self, age):
        self.age = age
    
    def greet(self):
        return f"Hello from Foo with age {self.age}"
```

Define Qux class

```python
class Qux:
    def __init__(self, message):
        self.message = message
    
    def forward(self):
        return f"Qux is moving forward with message: {self.message}"
```

Define Bar so it inherits from Foo and Qux

```python
class Bar(Foo, Qux):
    def __init__(self, age, extra, message):
        Foo.__init__(self, age)
        Qux.__init__(self, message)
        self.extra = extra
    
    def greet(self):
        return f"Hello from Bar with age {self.age} and extra {self.extra}"
    
    def use_forward(self):
        return self.forward()
```
